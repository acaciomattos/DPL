# MVP Scope

This repository implements the minimum viable path described across the attached materials.

## Included now

- baseline reproduction on a client-style frozen snapshot
- manual scenario editing through threshold overrides
- reusable derived-feature veto blocks
- persisted scenarios and results inside each study
- bounded recommendation search with three selectable modes
- Dash application shell aligned to the visual direction of the reference image

## Deferred from this first cut

- additional source adapters beyond the demonstrative Python adapter
- richer policy grammars and multi-block editing in the UI
- DuckDB-backed execution paths
- study creation workflows from uploaded files
- stronger uncertainty modeling such as temporal stability and reject-inference controls
- authentication, packaging and deployment automation
